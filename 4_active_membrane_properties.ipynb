## 1.1: Synaptic input <a name="1.1"></a>



Recall the following **key concepts**:

**Depolarization**: Neurons <ins>at rest</ins> have a <ins>membrane potential</ins> of approximately <ins>-70 mV</ins>, which we refer to as a <ins>polarized</ins> state (i.e., a state far from 0 mV). **Depolarizing currents** bring the membrane potential <ins>closer to 0 mV</ins>.

**Hyperpolarization**: In some cases, synaptic input results in a **hyperpolarizing effect**, driving the membrane potential even further away from 0 mV, thus making the neuron less likely to fire.

**EPSP** (**E**xcitatory **P**ost**s**ynaptic **P**otential): Synaptic input that has a depolarizing effect, moving the membrane potential closer to the threshold for an action potential.

**IPSP** (**I**nhibitory **P**ost**s**ynaptic **P**otential): Synaptic input that causes hyperpolarization, moving the membrane potential further from the firing threshold.

Whether a <ins>synaptic input is excitatory or inhibitory</ins> depends on the **neurotransmitter** (**NT**) released by the presynaptic neuron. According to **Dale's Principle**, a neuron releases the <ins>same neurotransmitter at all of its synapses throughout its lifetime</ins>, maintaining a consistent chemical action.

In the brain, the <ins>primary excitatory neurotransmitter</ins> is **glutamate**, while the <ins>main inhibitory neurotransmitter</ins> is **GABA** (gamma-Aminobutyric acid).

Note that so far all synaptic inputs in our models have been excitatory.

## 2.2: The Hodgkin-Huxley model <a name="2.2"></a>

The **model presented in Section 2.a** allows for indefinite depolarization of a neuron by linearly summing all EPSPs arriving at the soma. However, as we have seen, this behavior does not at all represent how real neurons function. When the membrane potential crosses a certain **depolarization threshold**, neurons generate a <ins>non-linear, all-or-nothing response</ins>: **The action potential**.

In 1952, Alan Hodgkin and Andrew Huxley developed a <ins>mathematical model</ins> that explains how action potentials are <ins>initiated and propagated</ins> along the squid giant axon.

$I_{\text{ion}} =
\underbrace{g_{L} (V_{m} - E_{L})}_{\text{Leak current, }I_{L}} +
\underbrace{\overline{g_{K}} *  n^4 (V{m} - E_{K})}_{\text{Potassium current, }I_{K}} +
\underbrace{\overline{g_{Na}} * m^3 * h (V_{m} - E_{Na})}_{\text{Sodium current, }I_{Na}}
$

The **total ionic current** $I_{ion}$ is a <ins>sum</ins> of the <ins>constant currents</ins> through so called **leak channels**, $I_{L}$ and the <ins>currents that are responsible for fast de- and repolarisation during an action potential</ins>: $I_{Na}$ and $I_{K}$. As we have seen in previous exercises, the current for each ion depends on **a)** the selective permeability of the membrane for the ion species (conductance, $g_{ion}$), and **b)** the difference between the membrane potential and the reversal potential of the ion species:

$I_{ion} = g_{ion} * (Vm - E_{ion})$

For several ions, the conductance $g_{ion}$ is not constant but modulated by a physical gate that is opened or closed with a certain probability. The probability for one of the ion channels to be open is given by $n_{ion}$. Thus, $n_{ion}$ can be interpreted as a proportion of the maximum conductance.

The actual ionic conductance can therefore be written as the maximal conductance $\overline{g_{ion}}$ multiplied by the gating variable so that:

$g_{ion}$= $\overline{g_{ion}}$  * $n_{ion}$

## Hodgkin–Huxley gating variables

### 1. Bounds of the gating variable

The gating variable $n_{\mathrm{ion}}$ represents the probability or proportion of open gates. Therefore,

$$
\boxed{0 \leq n_{\mathrm{ion}} \leq 1}
$$

Here, $n_{\mathrm{ion}}=0$ means that no gates are open, while $n_{\mathrm{ion}}=1$ means that all gates are open.

---

### 2. Probability that all gates are open

If a channel contains $x$ independent gates and each gate has an opening probability $n$, the probability that all $x$ gates are open is

$$
\boxed{P(\text{channel open})=n^x}
$$

For example, if the channel has four gates,

$$
P(\text{channel open})=n^4.
$$

---

### 3. Change in the proportion of open gates

The proportion of closed gates is $1-n$. Closed gates open with rate $\alpha_n$, while open gates close with rate $\beta_n$.

Therefore,

$$
\boxed{
\frac{dn}{dt}
=
\alpha_n(1-n)-\beta_n n
}
$$

The first term describes gates opening, and the second term describes gates closing.

---

### 4. Steady-state gating variable

At steady state, the proportion of open gates no longer changes:

$$
\frac{dn}{dt}=0.
$$

Therefore,

$$
0=\alpha_n(1-n_\infty)-\beta_n n_\infty.
$$

Solving for $n_\infty$ gives

$$
\boxed{
n_\infty
=
\frac{\alpha_n}{\alpha_n+\beta_n}
}
$$

---

### 5. Potassium current

The potassium current in the Hodgkin–Huxley model is

$$
\boxed{
I_K=\bar{g}_K n^4(V_m-E_K)
}
$$

The terms have the following meanings:

- $\bar{g}_K$ is the maximum potassium conductance.
- $n^4$ is the fraction of potassium channels that are open.
- $V_m-E_K$ is the electrochemical driving force for potassium.

The model assumes that every potassium channel contains four identical and independent activation gates. All four gates must be open for potassium ions to pass. Since the probability that one gate is open is $n$, the probability that all four gates are open is $n^4$.

The actual potassium conductance is therefore

$$
g_K=\bar{g}_K n^4.
$$

During depolarisation, $n$ increases relatively slowly. Potassium channels open, potassium generally flows out of the neuron, and the membrane repolarises towards the potassium reversal potential $E_K$.

---

### 6. Sodium current and inactivation

The sodium current is

$$
\boxed{
I_{\mathrm{Na}}
=
\bar{g}_{\mathrm{Na}}m^3h(V_m-E_{\mathrm{Na}})
}
$$

The terms have the following meanings:

- $\bar{g}_{\mathrm{Na}}$ is the maximum sodium conductance.
- $m^3$ describes sodium-channel activation.
- $h$ describes sodium-channel availability or inactivation.
- $V_m-E_{\mathrm{Na}}$ is the electrochemical driving force for sodium.

The model assumes that the sodium channel contains three activation gates represented by $m$ and one inactivation gate represented by $h$. The channel conducts only when all three activation gates are open and the inactivation gate is not blocking the channel.

Therefore, the fraction of conducting sodium channels is

$$
m^3h.
$$

During depolarisation, $m$ increases quickly and the sodium channels activate. Sodium enters the neuron and produces further depolarisation. Shortly afterwards, $h$ decreases and the channels become inactivated. This stops the sodium current even if the membrane remains depolarised.

During repolarisation, $m$ decreases and $h$ gradually recovers towards 1. The sodium channels then become available for a future action potential.

In summary,

$$
\boxed{
m=\text{activation},
\qquad
h=\text{availability/inactivation}
}
$$

**Solution:**
The refractory scale factor controls how strongly the refractory period is extended or adapted after the neuron fires.

A larger refractory scale factor produces longer refractory periods. This reduces the maximum firing rate and causes more incoming inputs to be ignored.

A smaller refractory scale factor produces shorter refractory periods. The neuron recovers sooner and can produce more spikes in response to high-frequency input.

Therefore, the refractory mechanism prevents the neuron from firing arbitrarily fast and limits the conversion of input frequency into output spike frequency.

> **Assignment 9**
>
> Does this model implement $\tau$? If so, how can you tell?

**Solution:**
Yes, the leaky integrate-and-fire model implements the membrane time constant $\tau$.

The membrane potential decays towards the resting potential according to

$$
\frac{dV_m}{dt}
=
-\frac{V_m-V_{\mathrm{rest}}}{\tau}.
$$

The solution has the form

$$
V_m(t)-V_{\mathrm{rest}}
=
\left(V_m(0)-V_{\mathrm{rest}}\right)e^{-t/\tau}.
$$

This can be observed in the interactive plot because an EPSP gradually decreases back towards the resting potential instead of remaining constant.

A small $\tau$ produces a fast decay, whereas a large $\tau$ produces a slow decay. Therefore, changing the time-constant slider changes how long an EPSP remains visible.

> **Assignment 10**
>
> How does this influence the neuron’s ability to integrate information?

**Solution:**
Increasing the membrane time constant $\tau$ slows the decay of each EPSP.

With a small time constant, the first EPSP has mostly disappeared before the next input arrives. The neuron therefore has difficulty adding the inputs together.

With a larger time constant, each EPSP persists for longer. If another input arrives before the previous EPSP has decayed, the two EPSPs sum through temporal summation.

Therefore,

$$
\boxed{
\text{larger }\tau
\Rightarrow
\text{longer integration window}
\Rightarrow
\text{stronger temporal summation}
}
$$

A larger $\tau$ consequently makes it more likely that multiple inputs will drive the membrane potential across the firing threshold. A smaller $\tau$ makes the neuron more sensitive to coincident inputs and less sensitive to inputs that are separated in time.

> **Assignment 12**
>
> Increase the distance to soma to 1mm and then to 2mm. What changes? When comparing your observations with the results you expect from cable theory, which parameter is not implemented in this model? How would the results change, if this parameter was implemented?  


**Solution:**
The model uses the distance to calculate a propagation delay:

$$
t_{\mathrm{delay}}=\frac{x}{v},
$$

where

$$
v=0.1\,\mathrm{mm/ms}.
$$

The different distances therefore produce the following delays:

| Distance | Delay |
|---:|---:|
| $0.1\,\mathrm{mm}$ | $1\,\mathrm{ms}$ |
| $1\,\mathrm{mm}$ | $10\,\mathrm{ms}$ |
| $2\,\mathrm{mm}$ | $20\,\mathrm{ms}$ |

Increasing the distance shifts the EPSP and any resulting spike towards later times. However, the amplitude of the EPSP remains unchanged in the interactive model.

The missing cable-theory parameter is the length constant $\lambda$.

In cable theory, the amplitude decreases with distance approximately according to

$$
\Delta V(x)=\Delta V(0)e^{-x/\lambda}.
$$

If $\lambda$ were implemented, an input from a more distant synapse would not only arrive later but would also be smaller at the soma.

Therefore:

- At $0.1\,\mathrm{mm}$, the EPSP would undergo little attenuation and could cross the threshold.
- At $1\,\mathrm{mm}$, the EPSP would arrive later and with a smaller amplitude.
- At $2\,\mathrm{mm}$, it would arrive even later and be attenuated further, possibly remaining below the firing threshold.

The current model implements distance-dependent delay but not distance-dependent attenuation:

$$
\boxed{
\text{greater distance}
\Rightarrow
\text{later arrival, but unchanged amplitude}
}
$$

A model implementing $\lambda$ would instead produce

$$
\boxed{
\text{greater distance}
\Rightarrow
\text{later arrival and smaller amplitude}
}
$$

In order to understand the power of signal integration in single neurons, we will briefly introduce the concept of coding (precisely rate and temporal coding) here. You will hear more about these concepts during the lecture, so for now it is sufficient to keep it simple.

It is in general a difficult task to assign meaning to the spiking dynamics of neurons in the brain. What does the neuron encode when it fires so and so often or at these specific times? Two common coding strategies are well described and here we introduce them briefly:

**Rate coding**: The spiking frequency of a neuron is informative about some property of the signals. For example, a sensory neuron will fire action potentials at higher frequencies, when the input stimulus is stronger.

**Temporal coding**: The firing frequency itself is not informative about the encoded information but rather temporal factors like the relative timing of single spikes.

> **Assignment 14**
>
> Does any of the distributions influence the neuron's output more than the others? Is there a particular distribution that critically influences the neuron's response so that - if not above / below a certain threshold - the neuron won't fire?

**Solution:**

The distributions do not all influence the neuron equally.

### Synaptic weight

The weight distribution determines the voltage contribution of each input. Larger weights produce larger EPSPs and make firing more likely.

If all weights are too small, even synchronised inputs may not produce enough depolarisation to reach the threshold.

### Synchronisation

The synchronisation distribution determines how closely together the inputs arrive.

High synchronisation causes many EPSPs to overlap and sum. Low synchronisation spreads them over time, allowing earlier EPSPs to decay before later ones arrive.

Synchronisation therefore has a particularly strong effect when the membrane time constant is short.

### Input rate

A higher input rate provides more events per unit time. This increases the probability that several inputs overlap and reach the firing threshold.

However, a high input rate does not guarantee firing if the weights are very small or the inputs are poorly synchronised.

### Synaptic distance

Distance changes when an input reaches the soma. A distribution of distances therefore spreads the arrival times of the inputs and can reduce their effective synchronisation.

In this model, distance does not reduce the amplitude of an EPSP because the length constant $\lambda$ is not implemented.

### Critical parameters

The most critical parameters are normally the synaptic weight and synchronisation.

The neuron fires only if the sum of the remaining EPSPs reaches the threshold:

$$
V_{\mathrm{rest}}
+
\sum_i
w_i e^{-(t-t_i)/\tau}
\geq
V_{\mathrm{threshold}}.
$$

If the weights are too small, or if synchronisation is too low, this condition is never fulfilled.

Therefore, synaptic weight determines whether enough total depolarisation is available, while synchronisation determines whether that depolarisation is available at the same time.

}
$$

The number of output spikes during a fixed interval then encodes the input rate.

> Find parameter regimes so that the neuron implements temporal coding.



**Solution:**
Temporal coding means that the precise timing of inputs carries information, rather than only their average rate.

A possible parameter regime for temporal coding is:

| Parameter | Example setting |
|---|---:|
| Input-rate mean | constant |
| Input-rate standard deviation | small |
| Distance mean | $1\,\mathrm{mm}$ |
| Distance standard deviation | small |
| Synaptic-weight mean | approximately $2$–$5\,\mathrm{mV}$ |
| Synaptic-weight standard deviation | small |
| Time constant | short, approximately $5$–$10\,\mathrm{ms}$ |
| Threshold | sufficiently high to require several overlapping inputs |

The input rate and total number of inputs should remain approximately constant. The synchronisation parameters are then changed.

With high synchronisation, many inputs arrive within a short interval. Their EPSPs overlap, causing a spike.

With low synchronisation, the same number of inputs is distributed over a longer interval. Earlier EPSPs decay before later EPSPs arrive, so the neuron may not spike.

Therefore,

$$
\boxed{
\text{same number of inputs}
+
\text{different timing}
\Rightarrow
\text{different neuronal response}
}
$$

This is temporal coding because the response depends on when the inputs arrive, not only on how many inputs arrive.

> **Assignment 17**

> For different parameter regimes (e.g. high input rate with poor synchronization), how many inputs are required to provoke an action potential at $t=20\text{ms}$?


**Solution:**

The given parameters are:

$$
\text{weight}=2.5\,\mathrm{mV}\pm2.5\,\mathrm{mV},
$$

$$
\tau=10\,\mathrm{ms},
\qquad
V_{\mathrm{threshold}}=-20\,\mathrm{mV}.
$$

The neuron begins at

$$
V_{\mathrm{rest}}=-70\,\mathrm{mV}.
$$

It therefore requires a total depolarisation of

$$
\Delta V_{\mathrm{required}}
=
-20-(-70)
=
50\,\mathrm{mV}.
$$

For inputs arriving at times $t_i$, the remaining contribution at $t=20\,\mathrm{ms}$ is approximately

$$
\Delta V(20)
=
\sum_i
w_i e^{-(20-t_i)/\tau}.
$$

An action potential occurs if

$$
\boxed{
\sum_i
w_i e^{-(20-t_i)/10}
\geq
50\,\mathrm{mV}
}
$$

### Perfectly synchronised inputs

If all inputs arrive at $t=20\,\mathrm{ms}$ and have the mean weight of $2.5\,\mathrm{mV}$, then

$$
N(2.5\,\mathrm{mV})\geq50\,\mathrm{mV}.
$$

Therefore,

$$
\boxed{N\geq20}
$$

Approximately 20 average-strength, perfectly synchronised inputs are required.

### Stronger inputs

If the relevant inputs have an average weight of $5\,\mathrm{mV}$, then

$$
N\geq\frac{50}{5}=10.
$$

Approximately 10 synchronised inputs are required.

### Inputs arriving 5 ms earlier

An input arriving at $15\,\mathrm{ms}$ has decayed for 5 ms by $t=20\,\mathrm{ms}$:

$$
2.5e^{-5/10}
\approx1.52\,\mathrm{mV}.
$$

The required number is then approximately

$$
N\geq\frac{50}{1.52}\approx33.
$$

### Inputs arriving 10 ms earlier

An input arriving at $10\,\mathrm{ms}$ contributes only

$$
2.5e^{-10/10}
\approx0.92\,\mathrm{mV}
$$

at $t=20\,\mathrm{ms}$.

The required number would therefore be approximately

$$
N\geq\frac{50}{0.92}\approx55.
$$

Since the model has only 20 synapses, inputs arriving this early cannot normally trigger a spike by themselves.

### Interpretation of different regimes

| Parameter regime | Expected result |
|---|---|
| High synchronisation and strong weights | Few inputs are required |
| High synchronisation and average weights | Approximately 20 inputs are required |
| Poor synchronisation | More than 20 inputs may be required |
| Inputs arriving much earlier | Their contribution decays and may be insufficient |
| High input rate with poor synchronisation | Many total inputs occur, but only inputs close to 20 ms contribute strongly |

Because the weights and input timings are randomly sampled, the exact number differs between runs. The appropriate method is to count the inputs arriving near $t=20\,\mathrm{ms}$ and consider both their weights and how much they have decayed.

The main conclusion is:

$$
\boxed{
\text{effective input}
=
\text{synaptic weight}
\times
\text{temporal decay}
}
$$

The neuron fires when the sum of these effective inputs reaches the required $50\,\mathrm{mV}$ depolarisation.
